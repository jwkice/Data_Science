Import packages

```python
import pandas as pd
import altair as alt
```

Load and check avocado data

```python
avocado_prices = pd.read_csv("data/avocado_prices.csv") # Read csv
avocado_prices  #Display df
```

Plot average price vs date

```python
alt.data_transformers.disable_max_rows()    # Disable max rows error message to allow plotting of all data
```

```python
avocado_plot = alt.Chart(avocado_prices).mark_point().encode(   # Initialize scatter plot for avocado prices
    x=alt.X("Date"),                    # Set x-axis as Date column
    y=alt.Y("average_price")            # Set y-axis as average price column
        .title("Average Price")         # Set title of y-axis
).configure_axis(                       # Configure font sizes for axes
    labelFontSize=12,
    titleFontSize=12
).properties(                           # Set plot title
    title="Avocado Average Price"
)
avocado_plot                            # Display plot
```

Aggregate by week

```python

avocado_aggregate = avocado_prices.groupby("wk").mean(numeric_only=True).reset_index()  # Aggregate df by week column, 
                                                                                        # calculate aggregated averages
avocado_aggregate           # Display aggregated df
```

Plot average price by week

```python
aggregate_week_plot = alt.Chart(avocado_aggregate).mark_point().encode( # Initialize scatter plot for aggregated avocado prices
    x=alt.X("wk", title="Week"),                    # Set x-axis as week column
    y=alt.Y("average_price")                        # Set y-axis as average price column
        .title("Average Price (in US Dollars)")     # Set title of y-axis
        .scale(zero=False)                          # Set scale of y-axis to not start at zero
).configure_axis(                                   # Configure font sizes for axes
    labelFontSize=12,
    titleFontSize=12
).properties(                                       # Set plot title
    title="Avocado Average Price per Week"
)
aggregate_week_plot                                 # Display plot
```

Calculate total volume - sum of three volume columns - original avocado data frame

```python
avocado_prices_total_volume = avocado_prices.assign(    # Create new df with new column
    total_volume=avocado_prices.loc[                    # New column is named total_volume
        :, avocado_prices.columns.str.contains('volume')].sum(axis=1)   # Select all rows,
                                                                        # only columns where name contains volume,
                                                                        # sum each row, assign to new column 
)
avocado_prices_total_volume     # Display df
```

Aggregate updated dataset by week

```python
avocado_aggregate_2 = avocado_prices_total_volume.groupby("wk").mean(numeric_only=True).reset_index()   # Aggregate df by week column,
                                                                                                        # calculate aggregated averages
avocado_aggregate_2         # Display aggregated df
```

Plot aggregated total volume

```python
aggregate_total_volume_plot = alt.Chart(avocado_aggregate_2).mark_point().encode(   # Initialize scatter plot for aggregated data
    x=alt.X("wk", title="Week"),                        # Set x-axis as week column
    y=alt.Y("total_volume")                             # Set y-axis as total volume column
        .title("Average Total Volume (lbs)")            # Set title of y-axis
        .scale(zero=False)                              # Set y-axis to not start at zero
).configure_axis(                                       # Configure font sizes for axes
    labelFontSize=12,
    titleFontSize=12
).properties(                                           # Set plot title
    title="Avocado Average Total Volume per Week"
)
aggregate_total_volume_plot                             # Display plot
```

Load and display departure bay temperature data

```python
departure_bay_temperature = pd.read_csv("data/departure_bay_temperature.csv", skiprows=2)   # Read data,
                                                                                            # skip two rows of metadata
departure_bay_temperature           # Display df
```

Use melt to reshape dataframe

```python
departure_bay_temperature = departure_bay_temperature.melt(     # Modify df with melt function
    id_vars="Year",                 # Set year column as id variable, to be used as row identifier
    var_name="Month",               # Set month column as variable name, name to use for column holding old column names
    value_name="Temperature"        # Set value name as temperature, for column to hold 'melted' values
)
departure_bay_temperature           # Display modified df
```

Plot average temperature vs year for the month of November

```python
november_temp_plot = alt.Chart(         # Initialize scatter plot
    departure_bay_temperature[departure_bay_temperature["Month"] == "Nov"]).mark_point().encode(    # Select input from rows
                                                                                                    # where month is November
    x=alt.X("Year")                                                 # Set x-axis as year column
        .title("Year")                                              # Set title of x-axis
        .scale(zero=False),                                         # Set x-axis to not start at zero
    y=alt.Y("Temperature")                                          # Set y-axis as temperature column
        .title("Temperature (Celsius)")                             # Set title of y-axis
        .scale(zero=False)                                          # Set y-axis to not start at zero
).configure_axis(                                                   # Configure font sizes of axes
    labelFontSize=12,
    titleFontSize=12
).properties(                                                       # Set plot title
    title="Departure Bay Temperature Monthly Average (November)"
)
november_temp_plot                                                  # Display plot
```

Plot average temp vs year for all months, in faceted display

```python
all_temp_plot = alt.Chart(departure_bay_temperature).mark_point().encode(   # Initialize scatter plot
    x=alt.X("Year")                                                         # Set x-axis as year column
        .title("Year")                                                      # Set title of x-axis
        .scale(zero=False),                                                 # Set x-axis to not start at zero
    y=alt.Y("Temperature")                                                  # Set y-axis as temperature column
        .title("Temperature (Celsius)")                                     # Set title of y-axis
        .scale(zero=False)                                                  # Set y-axis to not start at zero
).facet(                                                                    # Plot subsets (facets) of the data in separate subplots
    "Month",                                                                # Facet data by month value
    columns=4                                                               # Plot facets in four columns
).configure_axis(                                                           # Configure font sizes of axes
    labelFontSize=12,
    titleFontSize=12
).properties(                                                               # Set plot title
    title="Departure Bay Temperature Monthly Average (All Months)"
)
all_temp_plot                                                               # Display plot
```
