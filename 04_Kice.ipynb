```python
# Import libraries
import pandas as pd
import altair as alt
```

## Part 1: World Vaccination

Visualizing global vaccination rates for Polio and Hepatitis B using WHO data.

```python
# Task 1: Read and filter world_vaccination.csv
world_vaccination_raw = pd.read_csv("data/world_vaccination.csv")   # Read CSV
world_vaccination = world_vaccination_raw[world_vaccination_raw["who_region"] != "(WHO) Global"]    # Filter out rows with (WHO) Global
world_vaccination
```

```python
# Task 2: Scatter plot of % vaccinated vs year for all regions
scatter_vaccination = alt.Chart(world_vaccination).mark_point().encode( # Initialize scatter plot
    x=alt.X("yr").title("Year").scale(zero=False),              # X-axis = year, not starting from zero
    y=alt.Y("pct_vaccinated").title("Percentage Vaccinated")    # Y-axis = percent vaccinated
).configure_axis(                                               # Set size of axis text
    labelFontSize=12,
    titleFontSize=12
).properties(                                                   # Set plot title
    title="Vaccination Percentage by Year"
)
scatter_vaccination
```

```python
# Task 3: Add color and shape encoding for vaccine
scatter_vaccination_enhanced = alt.Chart(world_vaccination).mark_point().encode(    # Initialize scatter plot
    x=alt.X("yr").title("Year").scale(zero=False),                  # X-axis = year, not starting from zero
    y=alt.Y("pct_vaccinated").title("Precentage Vaccinated"),       # Y-axis = percent vaccinated
    color=alt.Color("vaccine").title("Vaccine Type"),               # Color = vaccine
    shape=alt.Shape("vaccine").title("Vaccine Type")                # Shape = vaccine
).configure_axis(                                                   # Set size of axis text
    labelFontSize=12,
    titleFontSize=12
).configure_legend(                                                 # Set size of legend text
    titleFontSize=12,
    labelFontSize=12
).properties(                                                       # Set plot title
    title="Vaccination Percentage by Year and Vaccine Type"
)
scatter_vaccination_enhanced
```

```python
# Task 4: Data frame for polio only
polio = world_vaccination[world_vaccination["vaccine"] == "polio"]  # Filter dataset to rows where vaccine is polio
polio
```

```python
# Task 5: Scatter plot for polio by region
polio_regions = alt.Chart(polio).mark_point().encode(           # Initialize scatter plot
    x=alt.X("yr").title("Year").scale(zero=False),              # X-axis = year, not starting from zero
    y=alt.Y("pct_vaccinated").title("Percentage Vaccinated"),   # Y-axis = percent vaccinated
    color=alt.Color("who_region").title("Region"),              # Color = region
    shape=alt.Shape("who_region").title("Region")               # Shape = region
).configure_axis(                                               # Set size of axis text
    labelFontSize=12,
    titleFontSize=12
).properties(                                                   # Set plot title and width
    title="Polio Vaccination Percentage by Year and Region",
    width=500
)
polio_regions
```

```python
# Task 6: Line plot for polio by region
polio_regions_line = alt.Chart(polio).mark_line().encode(       # Initialize line plot
    x=alt.X("yr").title("Year").scale(zero=False),              # X-axis = year, not starting from zero
    y=alt.Y("pct_vaccinated").title("Percentage Vaccinated"),   # Y-axis = percent vaccinated
    color=alt.Color("who_region").title("Region")               # Color = region
).configure_axis(                                               # Set size of axis text
    labelFontSize=12,
    titleFontSize=12
).properties(                                                   # Set plot title and width
    title="Polio Vaccination Percentage by Year and Region",
    width=500
)
polio_regions_line
```

```python
# Task 7: Facet line plot for polio and hepatitis B side-by-side
side_by_side_world = alt.Chart(world_vaccination).mark_line().encode(   # Initialize line plot
    x=alt.X("yr").title("Year").scale(zero=False),              # X-axis = year, not starting from zero
    y=alt.Y("pct_vaccinated").title("Percentage Vaccinated"),   # Y-axis = percent vaccinated
    color=alt.Color("who_region").title("Region")               # Color = region
).facet(                                                        # Facet by vaccine type
    "vaccine"
).configure_axis(                                               # Set size of axis text
    labelFontSize=12,
    titleFontSize=12
).properties(                                                   # Set plot title
    title="Vaccination Percentage by Year and Region"
)
side_by_side_world
```

```python
# Task 8: Facet line plot for polio and hepatitis B vertically
vertical_world = alt.Chart(world_vaccination).mark_line().encode(   # Initialize line plot
    x=alt.X("yr").title("Year").scale(zero=False),              # X-axis = year, not starting from zero
    y=alt.Y("pct_vaccinated").title("Percentage Vaccinated"),   # Y-axis = percent vaccinated
    color=alt.Color("who_region").title("Region")               # Color = region
).facet(                                                        # Facet by vaccine type, arrange in one column
    "vaccine",
    columns=1
).configure_axis(                                               # Set size of axis text
    labelFontSize=12,
    titleFontSize=12
).properties(                                                   # Set plot title
    title="Vaccination Percentage by Year and Region"
)
vertical_world
```

## Part 2: Fast-Food Chains in the United States

Analyzing the prevalence and distribution of fast-food restaurants on the West Coast.

```python
# Task 1: Read fast_food.csv
fast_food = pd.read_csv("data/fast_food.csv")   # Read CSV
fast_food
```

```python
# Task 2: Top 9 restaurants by location count in CA, WA, OR
west_coast = fast_food[fast_food["st"].isin(["CA", "WA", "OR"])]   # Filter to West Coast states
top_restaurants = west_coast["name"].value_counts().head(9).reset_index()   # Get top 9 restaurant names and counts
top_restaurants
```

```python
# Task 3: Bar chart of counts for top 9 fast food restaurants
count_bar_chart = alt.Chart(top_restaurants).mark_bar().encode(   # Initialize bar chart
    x=alt.X("name").title("Restaurant"),                 # X-axis = restaurant name
    y=alt.Y("count").title("Count on West Coast")        # Y-axis = count of locations
).configure_axis(                                        # Set size of axis text
    labelFontSize=12,
    titleFontSize=12
).properties(                                            # Set plot title
    title="Top 9 Fast Food Restaurants on the West Coast"
)
count_bar_chart
```

```python
# Task 4: Rotate x-axis labels for readability
count_bar_chart_A = count_bar_chart.configure_axisX(   # Rotate x-axis labels
    labelAngle=-45
)
count_bar_chart_A
```

```python
# Task 5: Swap axes for horizontal bar chart
count_bar_chart_B = count_bar_chart.encode(   # Swap axes for horizontal bar chart
    x=alt.X("count").title("Count on West Coast"),      # X-axis = count of locations
    y=alt.Y("name").title("Restaurant")                 # Y-axis = restaurant name
)
count_bar_chart_B
```

```python
# Task 6: West Coast state counts for top 9 restaurants
top_nine = west_coast[west_coast["name"].isin(top_restaurants["name"])]   # Filter to top 9 restaurants only
state_counts = top_nine["st"].value_counts().reset_index()   # Count restaurants by state
state_counts
```

```python
# Task 7: Bar plot of restaurant count by state
state_counts_plot = alt.Chart(state_counts).mark_bar().encode(   # Initialize bar chart
    x=alt.X("st").title("State"),                       # X-axis = state
    y=alt.Y("count").title("Number of restaurants")     # Y-axis = number of restaurants
).configure_axis(                                       # Set size of axis text
    labelFontSize=12,
    titleFontSize=12
).properties(                                           # Set plot title
    title="Count of Top 9 Fast Food Restaurants by State (West Coast)"
)
state_counts_plot
```

```python
# Task 8: Data frame for top_n_state
top_n_state = top_nine.value_counts(["st", "name"]).reset_index()   # Count combinations of state and restaurant
top_n_state
```

```python
# Task 9: Stacked bar chart of top 9 restaurants by state (normalized)
top_n_state_bar_stacked = alt.Chart(top_n_state).mark_bar().encode(   # Initialize bar chart
    x=alt.X("st").title("State"),                            # X-axis = state
    y=alt.Y("count").stack("normalize").title("Percentage"), # Y-axis = percentage of restaurants
    color=alt.Color("name", title="Restaurant")              # Color = restaurant name
).configure_axis(                                            # Set size of axis text
    labelFontSize=12,
    titleFontSize=12
).configure_legend(                                          # Set size of legend text
    titleFontSize=12,
    labelFontSize=12
).properties(                                                # Set plot title
    title="Top 9 Fast Food Restaurants by State (Normalized Stacked Bar)"
)
top_n_state_bar_stacked
```

```python
# Task 10: Faceted bar chart of counts for top 9 restaurants by state
top_n_state_bar_faceted = alt.Chart(top_n_state).mark_bar().encode(   # Initialize bar chart
    x=alt.X("name").title("Restaurant"),            # X-axis = restaurant name
    y=alt.Y("count").title("Restaurant count"),     # Y-axis = restaurant count
    color=alt.Color("name").title("Restaurant")     # Color = restaurant name
).facet(                                            # Facet by state
    "st"
).configure_axis(                                   # Set size of axis text
    labelFontSize=12,
    titleFontSize=12
).configure_legend(                                 # Set size of legend text
    titleFontSize=12,
    labelFontSize=12
).properties(                                       # Set plot title
    title="Top 9 Fast Food Restaurants by State (Faceted Bar Chart)"
)
top_n_state_bar_faceted
```
